```python
import pandas as pd

df = pd.read_csv(r"D:\survey\raw\df_export.csv")
df
```

```python
from pandasql import sqldf

query = "select row_number() OVER (ORDER BY timstamp) survey_key, * from df"

df = sqldf(query, locals())
df
```

```python
query = "select survey_key, python_skill, career, future_goal, expectations, pace, difficulty, balance  from df"

main_table = sqldf(query, locals())
main_table
```

```python
python_topics =  sqldf("select survey_key, python_topics  from df", locals())        # Each cell becomes a list
# python_topics
python_topics["python_topics"] = python_topics["python_topics"].str.split(",")          # Each cell becomes a list
python_topics = python_topics.explode("python_topics",ignore_index=True)

display(python_topics)
```
